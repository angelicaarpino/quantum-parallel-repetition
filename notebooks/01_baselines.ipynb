## Observations

* CHSH: `ω(CHSH^2) = 10/16 > (3/4)^2`, so classically joint play helps; quantumly the independent value `cos^4(π/8)` matches the NPA bound (perfect parallel repetition for XOR games).
* Feige: `ω = ω(G^2) = 1/2` (Feige's counterexample). Single copy: `ω* = 9/16` in dimension 3, certified by NPA 1+AB. For n = 2 the NPA bound is 1/2, so the independent quantum value `(9/16)^2 ≈ 0.316` is far from optimal.
* Magic square: `ω* = 1` for every n with 2 EPR pairs per copy; classically `ω(G^2) = 66/81 > (8/9)^2`.

## Week 5 starting point: joint strategy for Feige^2

See-saw on the repeated game with one shared state of local dimension 4 (lower bound).

```python
import warnings
warnings.filterwarnings('ignore', message='Solution may be inaccurate')
from qpr import feige, parallel, seesaw

g2 = parallel(feige(), 2)
res = seesaw(g2, 4, restarts=3, seed=0)
print(f'joint, d = 4: {res.value:.6f}   independent: {(9/16)**2:.6f}   NPA bound: 0.5')
```
