# Magic square

**Rules.** Uniform row x and column y; Alice fills the row with even parity, Bob the column with odd parity; win iff they agree on the shared cell (Mermin 1990, Peres 1990; Cleve et al. 2004).

* `ω(G^n)`: exact classical value; `ω(G)^n`: independent classical play
* quantum, independent: best single-copy strategy on each copy (lower bound)
* NPA: upper bound on `ω*(G^n)`
* joint n = 2: see-saw on `G^2` with one shared state (lower bound)
