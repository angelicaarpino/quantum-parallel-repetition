# CHSH (benchmark)

**Rules.** Uniform bits x, y; win iff a XOR b = x AND y (Clauser et al. 1969; Cleve et al. 2004).

* `ω(G^n)`: exact classical value; `ω(G)^n`: independent classical play
* quantum, independent: best single-copy strategy on each copy (lower bound)
* NPA: upper bound on `ω*(G^n)`
* joint n = 2: see-saw on `G^2` with one shared state (lower bound)

Results produced by `python experiments/chsh/run.py`.

## Observations

* Classically joint play helps: `ω(CHSH^2) = 10/16 > (3/4)^2`.
* Quantumly the independent value `cos^4(π/8)` equals the NPA bound: perfect parallel repetition for XOR games (Cleve et al. 2008). The joint see-saw in local dimension 4 finds the same value.
