# Feige's game

**Rules.** Uniform bits x, y; answers {0, 1, ⊥}; win iff (a, b) = (⊥, x) or (y, ⊥) (Feige 1991, in the form of Schmidt et al., arXiv:2510.08484).

* `ω(G^n)`: exact classical value; `ω(G)^n`: independent classical play
* quantum, independent: best single-copy strategy on each copy (lower bound)
* NPA: upper bound on `ω*(G^n)`
* joint n = 2: see-saw on `G^2` with one shared state (lower bound)

Results produced by `python experiments/feige/run.py`.

## Observations

* `ω(G) = ω(G^2) = 1/2`: Feige's counterexample to perfect parallel repetition.
* Single copy: `ω* = 9/16` in local dimension 3, certified by NPA 1+AB.
* n = 2: the NPA bound is 1/2, so independent quantum play `(9/16)^2 ≈ 0.316` is far from optimal. The joint see-saw reaches 1/2, which equals the classical joint value: joint play helps, but this is not a quantum advantage.
